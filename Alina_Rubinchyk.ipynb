```python
tier1_clicks = df[df["country_group"] == "Tier1"]["clicks"].sum()

print("Клики по Tier1:", tier1_clicks)
```

```python
breed1_clicks = df[df["breed"] == "breed_1"]["clicks"].sum()
breed1_clicks
```

```python
br1_impress = df[df["breed"] == "breed_1"]["impressions"].sum()
br1_impress
```

```python
fem_costs= df[df['audience'] == "Female"]["costs"].sum()
fem_costs
```

```python
male_costs= df[df['audience'] == "Male"]["costs"].sum()
male_costs
```

```python
result= round(male_costs/fem_costs,2)
result
```

```python
grouped = df.groupby("breed")[["users", "costs"]].sum()
```

```python
grouped["efficiency"] = grouped["users"] / grouped["costs"]
```

```python
top2 = grouped.sort_values("efficiency", ascending=False).head(2)

print(top2)
```

```python

grouped = df.groupby(["breed", "audience", "country_group"])[["costs", "users"]].sum()

grouped = grouped[grouped["users"] > 0]

grouped["CPA"] = grouped["costs"] / grouped["users"]
best_combos = grouped.sort_values("CPA").head(5)

print(best_combos)
```

```python

grouped = df.groupby(["breed", "audience"])[["clicks", "impressions"]].sum()

grouped["CTR"] = (grouped["clicks"] / grouped["impressions"]) * 100

ctr_df = grouped.reset_index()

print(ctr_df)
```

```python
tier1 = df[df["country_group"] == "Tier1"]
grouped = tier1.groupby("breed")[["clicks", "impressions"]].sum()
grouped["CTR"] = (grouped["clicks"] / grouped["impressions"]) * 100
top_breed = grouped.sort_values("CTR", ascending=False).head(1)

print(top_breed)
```
